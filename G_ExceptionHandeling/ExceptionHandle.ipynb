## Understanding the Exception Handeling
Exception Handeling in python is used to handle the errors occered     
and Take Correct Action without stoping Execution Of Program

## What is Exception Handelling 
Exception Handeling are Events that Disrupts the NormalFlow of Program   
They Occered When the Error is Encountered During Program Execution 

## Common Exceptions
1 . FileNotFound : Generally occors When File Is Not Found \
2 . zeroDivisionError : When The Value Is Divided By Zero \
3 . ValueError : Invalid Value  
4 . TypeError : Invalid Type  

```python
# a=b this will throw an error 
try:
    a=b
except:
    print(f" B is Not Defined")
```

```python
try:
    a=b
except NameError as nm: # Using KeyWord From Exception Class also We can Handle
    print(nm)
```

```python
try:
    result=10/0
except ZeroDivisionError as zdr:
    print(zdr)
    print("We Can Not Divide By Zero")
```

```python
try:
    n=int(input("Enter A Number"))
    result=10/n
except ZeroDivisionError as zdr:
    print(zdr)
except ValueError as ve:
    print(ve)
except Exception as ex:
    print(ex)
else:
    print(result)
finally:
    print("Execution is Completed")
```

```python
try:
    file=open("Example1.txt",'r')
    constant=file.read()
except FileNotFoundError as zdr:
    print(zdr)
except Exception as ex:
    print(ex)
else:
    print(constant)
finally:
    if 'file' in locals() or not file.closed():
        file.close()
        print("File Is Closed")
```
