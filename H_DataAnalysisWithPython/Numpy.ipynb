```python
## Performing Statistics Operation
arr2=np.array([1,2,3,4,5,6,7,8])
mean=np.mean(arr2)
print("The Mean Is :",mean)
median=np.median(arr2)
print("The Median Is :",median)

# mode=np.mod(arr2)
# print("The Mode Is :",mode)
std_dev=np.std(arr2)
round_std=np.round(std_dev,2) ##round decimal to 2
print("The Standard Deviation is :",round_std)

normalized_data=(mean)/std_dev
print("The Normalized Data ",normalized_data)

## The Variance 
variance=np.var(arr2)
print("The Variance is ",variance)

arr2[(arr2>=5) & (arr2<=8)]

arr_min=np.min(arr2)
arr_max=np.max(arr2)
print(f"The Min is : {arr_min} The MAx is : {arr_max}")


reverse_ele=arr2[::-1]
print("The Reverse array",reverse_ele)
```
