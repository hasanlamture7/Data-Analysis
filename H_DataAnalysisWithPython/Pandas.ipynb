```python
import pandas as pd
num=[1,2,3,4,5]

series_pd=pd.Series(num)
print(series_pd)
print(type(series_pd))
```

```python

#If we want to define own Index
dictionary_pd={"A":1,"B":2,"C":3,"D":4}
new_series=pd.Series(dictionary_pd)
print(new_series)
```

```python
data=[1,2,3,4,5]
index=["A","B","C","D","E"]
new_series=pd.Series(data,index=index)
print(new_series)
```

```python
data={
    "Name":["Hasan","Ibrahim","Maryam"],
    "Age":[23,60,50],
    "City":["Ajara","Kolhapur","Anaf"],
    "Gender":["M","M","F"]
}

df=pd.DataFrame(data)
print(df)
```
