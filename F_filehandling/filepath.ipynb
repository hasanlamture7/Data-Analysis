```python
path='example1.txt'
if os.path.exists(path):
    print(f"The Path '{path}' Exists ")
else:
    print(f"The Path '{path}' Does Not Exists")
```

```python
##Checking Wetehr the Path is  Directory or File 

path='example.txt'
if os.path.isfile(path):
    print(f"The Path '{path}' Is a File ")
elif os.path.isdir(path):
    print(f"The Path '{path}' Is a File ")
else:
    print(f"The Path '{path}' Nither a File Nor a Directory")
```

```python
##This way we can get the Absolute Path


reletive_path="example.txt"
absolute_path=os.path.abspath(reletive_path)
print(absolute_path)
```
