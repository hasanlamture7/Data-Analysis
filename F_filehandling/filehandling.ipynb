```python
## reading file line by line
with open("example.txt",'r') as file:
    for lines in file:
        print(lines.strip())
```

```python
## Append Over file without overwitting the file
with open('example.txt','a')as file:
    file.write("Hey I am appended \n")
```

```python
##writeing multipe line to a file
lines=["This is File \n","Which is Used Multiple \n","With The help of Python"]
with open('example.txt','a') as files:
    files.writelines(lines) ##writelines is used for adding multiple lines in file at onces
```

```python
## working with binary file bins
with open('example.bin','rb') as file:
    content=file.read()
    print(content)
```

```python
## writing to bin 
lines=b'\x00\x01\x02\x03\x04\x05'
with open('example.bin','wb') as file:
    file.write(lines)
```

```python
## accessment for this lec : Count the Numbe of lines and words and char in a lines present in a file

def function_read(file_path):
    with open(file_path,'r') as file:
        lines=file.readlines()  # FIRST  Need To Read lines as usaul
        line_count=len(lines) # count the lines by simple function
        word_count=sum(len(line.split()) for line in lines) # count the word using the split function 
        char_count=sum(len(line) for line in lines) # counting the each character

    return line_count,word_count,char_count



file_path="example.txt"
lines,words,char=function_read(file_path)

print(f"Lines : {lines}, Words : {words}, char : {char}")

```
